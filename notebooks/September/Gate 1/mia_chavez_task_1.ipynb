## Source identity and access

```python
import os

import pandas as pd
```

```python
# Checks shape of the data
df = pd.read_csv('../../../data/allstate_claims_data.csv')
data_shape = df.shape

expected_shape = (188318, 132)

print("Data shape:", data_shape)

if data_shape != expected_shape:
    print(f"Data shape does not match expected: expected {expected_shape}, got {data_shape}", "\n")
else:
    print("Data shape matches :D", "\n")
```

```python
# Checks file size
file_size = os.path.getsize('../../../data/allstate_claims_data.csv')
expected_size = 70025339

print("File size:", file_size)

if file_size != expected_size:
    print(f"File size does not match expected: expected {expected_size}, got {file_size}", "\n")
else:
    print("File size matches :D", "\n")
```

# Schema and Integrity

```python
# Check header order and column roles
expected_header = ['id'] + [f'cat{i}' for i in range(1,117)] + [f'cont{i}' for i in range(1,15)] + ['loss'] 
if list(df.columns) != expected_header:
   print(f"Header order does not match expected: expected {expected_header}, got {list(df.columns)}", "\n")
else:
    print("Header order matches :D", "\n")
    
```

```python
# checks for missing values in id
values = df['id'].isna().sum()

if values > 0:
    print(f"Missing values found in id column: {values}", "\n")
else:
    print("No missing values found in id column :D", "\n")

# check if id column is unique
total_unique_ids = df['id'].nunique()
expected_unique_ids = expected_shape[0]
if total_unique_ids != expected_unique_ids:
    print(f"Number of unique IDs does not match expected: expected {expected_unique_ids}, got {total_unique_ids}", "\n")
else:
    print(f"Number of unique IDs matches expected :D", "\n")
```

```python
# checks for duplicate rows
unique_rows = df.drop_duplicates()
num_unique_rows = unique_rows.shape[0]
if num_unique_rows != expected_shape[0]:
    print(f"Number of unique rows does not match expected: expected {expected_shape[0]}, got {num_unique_rows}", "\n")
else:
    print(f"Number of unique rows matches expected :D", "\n")
```

```python
# checks for missing cells

df_missing_cells = df.isna().sum().sum()
if df_missing_cells > 0:
    print(f"Missing cells found in the dataset: {df_missing_cells}", "\n")
else:
    print("No missing cells found in the dataset :D", "\n")
```

```python
# checks that continuous columns are numeric
continuous_columns = [f'cont{i}' for i in range(1,15)]
non_numeric_columns = [col for col in continuous_columns if not pd.api.types.is_numeric_dtype(df[col])]
if non_numeric_columns:
    print(f"Non-numeric columns found in continuous columns: {non_numeric_columns}", "\n")
else:
    print("All continuous columns are numeric :D", "\n")
```

```python
# checks that loss is numeric, finite and strictly positive
if not pd.api.types.is_numeric_dtype(df['loss']):
    print("Loss column is not numeric", "\n")
elif not df['loss'].apply(lambda x: pd.notnull(x) and x > 0).all():
    print("Loss column contains non-finite or non-positive values", "\n")
else:
    print("Loss column is numeric, finite and strictly positive :D", "\n")
```

```python
# check that cat columns are categorical
cat_cols = [c for c in df.columns if c.startswith('cat')]

for c in cat_cols[:]:
    print(c, df[c].dtype, df[c].unique()[:5])
```

```python
# Builds a profile of the dataset
profile = pd.DataFrame({
    'dtype': df.dtypes.astype(str),
    'missing': df.isna().sum(),
    'n_unique': df.nunique(),
    'min': df.min(numeric_only=True),
    'max': df.max(numeric_only=True),
})

profile.to_csv('profile.csv')
profile
```
